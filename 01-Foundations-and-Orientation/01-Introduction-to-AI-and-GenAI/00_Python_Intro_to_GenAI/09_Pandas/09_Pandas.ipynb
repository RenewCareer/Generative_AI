## 1. Series and DataFrames


````python
import pandas as pd
import numpy as np

# Series — a single column
ages = pd.Series([25,30,22,28], index=["Alice","Bob","Charlie","Diana"])
print(ages)
print("Alice's age:", ages["Alice"])

# DataFrame — a full table
data = {
    "Name":   ["Alice","Bob","Charlie","Diana"],
    "Age":    [25, 30, 22, 28],
    "City":   ["Accra","Kumasi","Tamale","Accra"],
    "Score":  [85.0, 92.0, 78.0, 95.0],
}
df = pd.DataFrame(data)
print("\n", df)
print("Shape:", df.shape)
````

## 2. Exploring a Dataset


````python
import pandas as pd, numpy as np
np.random.seed(42)
n = 50
df = pd.DataFrame({
    "Name":  [f"S{i}" for i in range(n)],
    "Age":   np.random.randint(18,30,n),
    "Math":  np.random.randint(40,100,n).astype(float),
    "English": np.random.randint(40,100,n).astype(float),
    "City":  np.random.choice(["Accra","Kumasi","Tamale"],n),
})
df.loc[[5,15], "Math"] = np.nan  # add some missing values

print(df.head())          # first 5 rows
print("\n", df.tail(3))  # last 3 rows
df.info()
print("\n", df.describe())
print("\nMissing:", df.isnull().sum())
````

## 3. Selecting and Filtering


````python
# Select column(s)
print(df["Name"].head())
print(df[["Name","Math"]].head())

# By position (iloc) and label (loc)
print(df.iloc[0])           # first row
print(df.loc[0:2, ["Name","Math","City"]])  # rows 0-2, specific cols

# Filter rows by condition
high = df[df["Math"] > 80]
print(f"\nStudents with Math>80: {len(high)}")

# Multiple conditions
top = df[(df["Math"] > 70) & (df["English"] > 70)]
print(f"Math>70 AND English>70: {len(top)}")

# isin()
accra = df[df["City"].isin(["Accra","Kumasi"])]
print(f"From Accra or Kumasi: {len(accra)}")
````

## 4. Sorting, Missing Values, and GroupBy


````python
# Sort
print(df.sort_values("Math", ascending=False).head(5)[["Name","Math"]])

# Fill missing values
df["Math"] = df["Math"].fillna(df["Math"].median())
print("Missing after fill:", df["Math"].isnull().sum())

# GroupBy
print("\nMean Math by City:")
print(df.groupby("City")["Math"].mean().round(1))

print("\nFull summary by City:")
print(df.groupby("City").agg(
    Count=("Name","count"),
    Avg_Math=("Math","mean"),
    Avg_English=("English","mean")
).round(1))
````

## 5. Merging DataFrames


````python
import pandas as pd

students = pd.DataFrame({
    "id":   [1,2,3,4],
    "name": ["Alice","Bob","Charlie","Diana"],
    "city_id": [1,2,1,3]
})
cities = pd.DataFrame({
    "city_id": [1,2,3],
    "city":    ["Accra","Kumasi","Tamale"]
})

# Inner join — only matching rows
merged = pd.merge(students, cities, on="city_id")
print(merged)

# Concatenate — stack vertically
part1 = students.iloc[:2]
part2 = students.iloc[2:]
print("\nConcatenated:")
print(pd.concat([part1, part2], ignore_index=True))
````

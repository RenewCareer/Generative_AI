# For Loops
- Iterates over a sequence

```python
# List of number 
numbers = [10, 20, 30, 40, 50]

for number in numbers:
    print(number)
```

```python
# Using for loops to calculate sum

marks = [75, 80, 64, 90, 85]

total = 0

for mark in marks:
    total = total + mark

print("Total marks:", total)
```

```python
print(mark)
```

```python
total = 0
for mark in marks:
    total += mark

print(total)
```

```python
# Using for loops with strings
name = "Python" # String

for letter in name:
    print(letter)
```

```python
# Using for loops with dictionaries

student = {
    "name" : "Desmond",
    "marks" : 85,
    "course" : "Data Science"
}

for key, value in student.items():
    print(key, ":", value)
```

```python
# use for loops with range()
for i in range(5):
    print(i)
```

```python
for i in range(1, 6):
    print(i)
```

```python
for i in range(2, 12, 2):
    print(i)
```

```python
for i in range(5, 0, -1):
    print(i)
```

# While loops
A while loop repeats a block of code as long as the condition remains true

```python
count = 0

while count <= 5:
    print(count)
    count += 1
```

```python
print(count)
```

```python
it_count = 1

while it_count <= 4:
    print(f"iter count is {it_count}")
    it_count += 1
```

# break, pass and continue

```python
# Break 

for number in range(1, 10):
    if number == 5:
        break
    print(number)
```

```python
# Continue
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
```

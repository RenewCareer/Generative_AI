## 1. Syntax and Indentation

Python uses **indentation (spaces)** to define code blocks — not curly braces `{}`.
Standard = **4 spaces** per level.


```python
print("hello, world!")
```

```python
# Correct indentation
if True:
    print("Inside the if block")   # 4 spaces
    print("Still inside")           # 4 spaces
print("Outside the if block")       # 0 spaces
```

# Data Types in Python
1. Integers - Whole Number
2. Floats - Decimal Numbers
3. Strings - A sequence of characters - ""
4. Boolean - Logical data type - True or False

```python
name       = "Alice"     # str  → text
age        = 25          # int  → whole number
height     = 5.7         # float → decimal
is_student = True        # bool  → True or False

print(name, type(name))
print(age,  type(age))
print(height, type(height))
print(is_student, type(is_student))
```

### The four basic data types

| Type | Example | Use |
|------|---------|-----|
| `str` | `"Alice"` | Text |
| `int` | `25` | Whole numbers |
| `float` | `5.7` | Decimal numbers |
| `bool` | `True` | Yes/No flags |


## 3. Type Casting (Converting Between Types)


```python
age = "30"
```

```python
type(age)
```

```python
age_to_int = int(age)
```

```python
age_to_int
```

```python
type(age_to_int)
```

```python
# String → Integer
age_str = "30"
age_int = int(age_str)
print(age_int, type(age_int))

# Integer → Float
num_float = float(7)
print(num_float)

# Float → Integer (truncates — does NOT round)
print(int(9.99))   # → 9

# Number → String (needed to concatenate with text)
score = 95
print("Your score is: " + str(score))
```

## 4. Input and Output


```python
name = input("What is your name?")
print(name)
```

```python
# Inputs always returns a string
age = input("What is your age? ")
print(age)
type(age)
```

```python
# OUTPUT — the print() function
name = "Bob"
age  = 28

print("Name:", name, "Age:", age)              # comma-separated
print(f"My name is {name} and I am {age}.")    # f-string (recommended)

# Number formatting with f-strings
pi = 3.14159
print(f"Pi to 2 decimal places: {pi:.2f}")
print(f"Pi to 4 decimal places: {pi:.4f}")

# INPUT — uncomment to try interactively
# your_name = input("What is your name? ")
# print(f"Hello, {your_name}!")
```

## 5. Comments and Code Readability

Comments start with `#`. Python ignores them — they are for humans.


```python
# Single-line comment

"""
Multi-line comment (also called a docstring).
Use this to describe a function or explain a block.
"""

# Good comment: explains WHY
tax_rate = 0.2  # Ghana PAYE rate for this income bracket

# Bad comment: just restates the code (avoid)
x = x + 1  # add 1 to x    ← unhelpful
```

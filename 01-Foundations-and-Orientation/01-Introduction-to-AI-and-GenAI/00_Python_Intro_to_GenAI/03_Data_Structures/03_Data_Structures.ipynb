# Set Operations

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
```

```python
# Unions
print(A | B)
```

```python
# Alternative
print(A.union(B))
```

```python
# Intersection 
print(A & B)
```

```python
# Alternative 
print(A.intersection(B))
```

```python
# Set Difference
print(A - B)
```

```python
# Alternative
print(A.difference(B))
```

```python
# Symetric Difference
print(A ^ B)
```

```python
# Alternative 
print(A.symmetric_difference(B))
```

## Subset

```python
A = {1, 2}
B = {1, 2, 3, 4}

# Check
print(A.issubset(B))
```

```python
# Superset
print(B.issuperset(A))
```

```python
fruits_set = {"apple", "banana", "cherry", "apple"}  # duplicate removed!
print(fruits_set)

a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print("Union:       ", a | b)   # all items
print("Intersection:", a & b)   # common items
print("Difference:  ", a - b)   # in a but not b

# Very useful: remove duplicates from a list
names = ["Alice", "Bob", "Alice", "Charlie", "Bob"]
print("Unique:", list(set(names)))
```

## 6. Dictionaries

Stores **key: value** pairs. Use `{}` with colons.


```python
student = {
    "name" : "John",
    "age" : 25,
    "Course" : "Data Science"
}
```

```python
type(student)
```

### Characteristics
1. Mutable - Can change
2. Ordered - Maintain insertion order
3. Key-Value Structure
4. Unique keys
5. Fast lookup

```python
print(student)
```

```python
# Accessing value
# - get the name of the student
print(student["name"])
```

```python
# using the .get() method
# - access the age
print(student.get("age"))
```

```python
# If we try to access a key which does not exist, we get an error
print(student["email"])
```

```python
# A safer way
print(student.get("email")) # Return None
```

## Adding items to dictionaries

```python
print(student)
```

```python
student["email"] = "john@mail.com"
print(student)
```

```python
# Updating the value 
# Updating the age from 25 to 27
student["age"] = 27
print(student)
```

## Removing items

```python
# 1. using pop
student.pop("age")
print(student)
```

```python
# popitem - removes the last inserted item
student.popitem()
print(student)
```

```python
# del used to remove a particular key and value
del student["Course"]
```

```python
print(student)
```

```python
# clear remove everything
student.clear()
print(student)
```

```python
student = {
    "name": "Alice",
    "age": 22,
    "courses": ["Math", "CS", "Statistics"]
}

print(student["name"])          # Alice
print(student.get("gpa", "N/A")) # default if key missing

student["email"] = "alice@example.com"   # add new key
student["age"] = 23                      # update

# Loop over key-value pairs
for key, value in student.items():
    print(f"  {key}: {value}")

# Dictionary comprehension
squares = {x: x**2 for x in range(1, 6)}
print(squares)
```

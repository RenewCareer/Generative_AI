# Topic 12: EDA Workflow (Exploratory Data Analysis)

EDA = asking the right questions and letting the data answer.

**Steps:** Overview → Univariate → Bivariate → Multivariate → Insights


## 0. Build Dataset


```python
import numpy as np, pandas as pd
np.random.seed(0)
n = 200
df = pd.DataFrame({
    "Age":        np.random.randint(22,60,n),
    "Gender":     np.random.choice(["Male","Female"],n,p=[0.55,0.45]),
    "Education":  np.random.choice(["BSc","MSc","PhD","HND"],n,p=[0.45,0.30,0.10,0.15]),
    "Department": np.random.choice(["Engineering","Sales","HR","Finance"],n),
    "Experience": np.random.randint(0,30,n),
    "Salary":     np.random.normal(60000,18000,n).clip(20000,130000).round(-2),
    "Performance":np.random.choice([1,2,3,4,5],n,p=[0.05,0.10,0.30,0.35,0.20]),
})
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
print(df.shape)
df.head()
```

## 1. Dataset Overview


```python
df.info()
print(df.describe().round(1))
print("\nMissing:", df.isnull().sum())
```

## 2. Univariate Analysis — one variable at a time


```python
for col in ["Age","Salary","Experience"]:
    s = df[col]
    print(f"\n{col}: mean={s.mean():.1f}, median={s.median():.1f}, std={s.std():.1f}, skew={s.skew():.2f}")

for col in ["Gender","Education","Department"]:
    print(f"\n{col}:\n{df[col].value_counts()}")
```

## 3. Bivariate Analysis — two variables


```python
print("Salary by Gender:")
print(df.groupby("Gender")["Salary"].agg(["mean","median","count"]).round(0))

print("\nSalary by Department:")
print(df.groupby("Department")["Salary"].mean().sort_values(ascending=False).round(0))

print("\nCorrelation matrix:")
print(df[["Age","Experience","Salary","Performance"]].corr().round(3))
```

## 4. Multivariate Analysis


```python
print("Salary by Gender AND Department:")
pivot = df.pivot_table(values="Salary", index="Department", columns="Gender", aggfunc="mean")
print(pivot.round(0))

print("\nPerformance by Department:")
print(df.groupby("Department")["Performance"].mean().round(2).sort_values(ascending=False))
```

## 5. Key Insights


```python
print("=== KEY INSIGHTS ===")
print(f"Mean salary  : ${df['Salary'].mean():,.0f}")
print(f"Median salary: ${df['Salary'].median():,.0f}")
print(f"Salary skew  : {df['Salary'].skew():.2f} (right-skewed = a few high earners)")

gap = df.groupby("Gender")["Salary"].mean()
print(f"\nGender gap: Male ${gap.get('Male',0):,.0f} vs Female ${gap.get('Female',0):,.0f}")

top_dept = df.groupby("Department")["Salary"].mean().idxmax()
print(f"Top-paying dept: {top_dept}")

print(f"\nExp ↔ Salary corr: {df['Experience'].corr(df['Salary']):.3f}")
```

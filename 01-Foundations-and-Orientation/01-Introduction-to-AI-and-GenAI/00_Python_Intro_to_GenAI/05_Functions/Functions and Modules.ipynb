## Functions with Parameters

```python
first_name = "Desmond"
last_name = "Onam"
```

```python
first_name + last_name
```

```python
first_name + " " + last_name
```

```python
print(f"{first_name} {last_name}")
```

```python
price = 19.999999
print(f" Price is: ${price:.2f}")
```

```python
def greet_student(name):
    print(f"Hello, {name}!")
    # print("Hello, " + name + "!")
    print("Welcome to Python")
```

```python
greet_student("Desmond")
```

```python
greet_student("Faith")
```

```python
greet_student("Anil")
```

```python
# Write a function to take in the marks of 3 students and then calculates their average

def calculate_average(mark1, mark2, mark3):
    average = (mark1 + mark2 + mark3) / 3
    print(f"Average marks is: {average}")
```

```python
calculate_average(70, 80, 90)
```

```python
calculate_average(65, 73, 87)
```

```python
calculate_average(60, 75, 85)
```

### Returning results from functions
`return`

```python
def add_numbers(a, b):
    return a + b
```

```python
result = add_numbers(10, 20)
print(result)
```

```python
print(result * 2)
```

```python
result = add_numbers(4, 7)
print(result)
```

```python
print(result * 3)
```

```python
age = 15
print(age)
```

```python
print(age * 3)
```

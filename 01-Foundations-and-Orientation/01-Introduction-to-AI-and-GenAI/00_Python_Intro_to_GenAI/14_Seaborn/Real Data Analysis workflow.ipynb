# Step: Distribution Analysis

```python
# Revenue Distribution
sns.histplot(
    df['Revenue'],
    kde = True
)
```

```python
# Order Quantity distibution
sns.histplot(
    df['Order_Quantity'],
    kde = True
)
```

# step 4. Outlier Analysis


```python
sns.boxplot(
    x = df['Revenue']
)
```

```python
# Step 5: Correlation Analysis

corr = df.corr(numeric_only = True)

sns.heatmap(
    corr,
    annot = True,
    cmap = 'coolwarm'
)
```

## 1. Reading and Writing Text Files


```python
# Write a text file
with open("sample.txt", "w") as f:
    f.write("Line 1: Hello\n")
    f.write("Line 2: Python file handling\n")
    f.write("Line 3: Easy and clean\n")
print("File written.")

# Read entire file
with open("sample.txt", "r") as f:
    content = f.read()
print(content)

# Read line by line
with open("sample.txt", "r") as f:
    for line in f:
        print(">", line.strip())

# Append without overwriting
with open("sample.txt", "a") as f:
    f.write("Line 4: Appended later\n")
```

## 2. Working with CSV Files

CSV (Comma-Separated Values) is the most common format for datasets.


```python
import csv

# Write CSV
students = [
    ["Name", "Age", "Score"],
    ["Alice", 22, 85],
    ["Bob",   25, 92],
    ["Charlie", 21, 78],
]
with open("students.csv", "w", newline="") as f:
    csv.writer(f).writerows(students)
print("students.csv created.")

# Read with DictReader (each row is a dictionary)
with open("students.csv", "r") as f:
    for row in csv.DictReader(f):
        print(f"  {row['Name']} scored {row['Score']}") 
```

## 3. Working with JSON Files

JSON is used heavily in APIs and web services.


```python
import json

data = {
    "company": "TechCorp",
    "employees": [
        {"id": 1, "name": "Alice", "salary": 70000},
        {"id": 2, "name": "Bob",   "salary": 90000},
    ]
}

# Write JSON
with open("employees.json", "w") as f:
    json.dump(data, f, indent=4)
print("JSON saved.")

# Read JSON
with open("employees.json", "r") as f:
    loaded = json.load(f)
for emp in loaded["employees"]:
    print(f"  {emp['name']}: ${emp['salary']:,}")
```

```python
# An API lets your Python program talk to an online service.
# Most APIs return JSON. You call them with the requests library.

try:
    import requests
    url = "https://official-joke-api.appspot.com/random_joke"
    response = requests.get(url, timeout=5)
    if response.status_code == 200:
        joke = response.json()
        print("Setup:    ", joke["setup"])
        print("Punchline:", joke["punchline"])
except ImportError:
    print("Install requests: pip install requests")
except Exception as e:
    print(f"Could not reach API: {e}")
```

## ✅ Summary

| Task | How |
|---|---|
| Write text | `open(file, 'w')` + `f.write()` |
| Read text | `open(file, 'r')` + `f.read()` |
| Write CSV | `csv.writer(f).writerows(data)` |
| Read CSV | `csv.DictReader(f)` |
| Write JSON | `json.dump(data, f, indent=4)` |
| Read JSON | `json.load(f)` |
| Call API | `requests.get(url).json()` |

## 1. What is Deep Learning?

Deep Learning = ML with **multi-layer neural networks**.

```
Input → [Layer 1] → [Layer 2] → ... → [Layer N] → Output
```

- Classic ML: you engineer features manually
- Deep Learning: the network learns features automatically from raw data

**Excels at:** images, text, audio, video


## 2. How a Single Neuron Works

```
Inputs (x₁, x₂, x₃)
   × weights (w₁, w₂, w₃)
   + bias (b)
   → activation function
   → output
```

Exactly linear regression — but with an activation function and millions of neurons chained together.


````python
import math

def sigmoid(z): return 1 / (1 + math.exp(-z))
def relu(z):    return max(0, z)

def neuron(inputs, weights, bias, activation="sigmoid"):
    z = sum(w*x for w,x in zip(weights,inputs)) + bias
    return sigmoid(z) if activation=="sigmoid" else relu(z)

inputs  = [6.0, 0.85]   # [hours_studied, attendance_rate]
weights = [0.4, 0.6]
bias    = -0.5

out = neuron(inputs, weights, bias)
print(f"Output: {out:.4f} → Prediction: {'Pass' if out >= 0.5 else 'Fail'}")

print("\nActivation functions:")
for z in [-3,-1,0,1,3]:
    print(f"  z={z:3d}  sigmoid={sigmoid(z):.3f}  relu={relu(z):.1f}")
````

## 3. Understanding Transformers

Transformers (2017) power **all** modern GenAI: GPT, Claude, Gemini, BERT, LLaMA.

**Key innovation: Self-Attention**
> Each word looks at ALL other words to understand context.

| Component | Purpose |
|---|---|
| Tokenisation | Text → tokens |
| Embeddings | Tokens → vectors |
| Positional encoding | Adds word position info |
| Self-Attention | Each token attends to all others |
| Feed-Forward layers | Further processing |
| Stacking | 96+ layers deep in large models |


````python
# Conceptual self-attention demo
sentence = "The cat sat on the mat"
words = sentence.split()
focus_word = "sat"
print(f'Attention weights when processing "{focus_word}":')
import numpy as np; np.random.seed(7)
weights = np.array([0.05, 0.40, 0.15, 0.20, 0.05, 0.15])
for word, w in zip(words, weights):
    bar = "█" * int(w*40)
    print(f"  {word:8}: {w:.2f}  {bar}")
````

## 4. Preview: PyTorch Neural Network


````python
try:
    import torch, torch.nn as nn

    class SimpleNN(nn.Module):
        def __init__(self):
            super().__init__()
            self.fc1 = nn.Linear(4, 8)
            self.relu = nn.ReLU()
            self.fc2 = nn.Linear(8, 1)
            self.sigmoid = nn.Sigmoid()
        def forward(self, x):
            return self.sigmoid(self.fc2(self.relu(self.fc1(x))))

    model = SimpleNN()
    print(model)
    print(f"Parameters: {sum(p.numel() for p in model.parameters())}")

except ImportError:
    print("Install PyTorch: pip install torch")
    print("Structure would look like:")
    print("  fc1: Linear(4 → 8) + ReLU")
    print("  fc2: Linear(8 → 1) + Sigmoid")
````

## 2. Python as a Calculator

Python can do maths instantly — no extra setup needed.


````python
print(5 + 3)    # Addition
print(10 - 4)   # Subtraction
print(6 * 7)    # Multiplication
print(15 / 4)   # Division
print(15 // 4)  # Floor division (whole number)
print(15 % 4)   # Modulus (remainder)
print(2 ** 10)  # Power
````

## 5. Check Installed Libraries

Run the cell below to see which data science libraries are ready on your machine.


````python
libraries = ["numpy", "pandas", "matplotlib", "seaborn", "sklearn"]
for lib in libraries:
    try:
        __import__(lib)
        print(f"  {lib:15} -- INSTALLED ✓")
    except ImportError:
        print(f"  {lib:15} -- NOT FOUND  (pip install {lib})")
````

````python
# Sample Data
import pandas as pd

data = {
    "Month" : ['Jan', 'Feb', 'Mar', 'Apr', 'May'],
    "Sales" : [100, 120, 90, 140, 170],
    "Profit" : [20, 25, 15, 35, 50]
}

df = pd.DataFrame(data)

df
````

# Line Plot
- Used to see the trend over time

````python
plt.plot(df["Month"], df["Sales"])

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
````

# Customizing line plot

````python
plt.plot(
    df['Month'],
    df['Sales'],
    color = "red",
    linestyle = '--',
    marker = 'o',
    linewidth = 3
)
plt.show()
````

# Multiple Plots

````python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
np.random.seed(42)
print('Libraries loaded.')
````

## 1. Anatomy of a Plot

```
fig, ax = plt.subplots()  # fig = canvas, ax = the chart area
ax.plot(x, y)             # draw something
ax.set_title('My Plot')   # add title
ax.set_xlabel('X axis')   # label axes
plt.show()                # display
```


## 2. Line Plot


````python
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
sales_23 = [120,135,148,162,175,190,180,168,155,170,185,200]
sales_24 = [130,142,160,172,188,202,195,180,170,185,200,220]

fig, ax = plt.subplots(figsize=(10,4))
ax.plot(months, sales_23, marker="o", label="2023", linewidth=2)
ax.plot(months, sales_24, marker="s", label="2024", linewidth=2, linestyle="--")
ax.set_title("Monthly Sales 2023 vs 2024", fontsize=13, fontweight="bold")
ax.set_xlabel("Month"); ax.set_ylabel("Sales")
ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
````

## 3. Bar Chart


````python
depts = ["Engineering","Sales","HR","Finance","Marketing"]
salary = [85000, 65000, 58000, 72000, 60000]

fig, ax = plt.subplots(figsize=(8,4))
bars = ax.bar(depts, salary, color=["#2196F3","#FF9800","#4CAF50","#9C27B0","#F44336"])
for b in bars:
    ax.text(b.get_x()+b.get_width()/2, b.get_height()+500,
            f"${b.get_height():,.0f}", ha="center", fontsize=8)
ax.set_title("Average Salary by Department")
ax.set_ylabel("Salary ($)"); ax.grid(axis="y", alpha=0.3)
plt.tight_layout(); plt.show()
````

## 4. Histogram


````python
scores = np.random.normal(70, 12, 500)

fig, ax = plt.subplots(figsize=(8,4))
ax.hist(scores, bins=20, color="steelblue", edgecolor="white", alpha=0.8)
ax.axvline(scores.mean(),   color="red",  linestyle="--", label=f"Mean={scores.mean():.1f}")
ax.axvline(np.median(scores), color="navy", linestyle=":",  label=f"Median={np.median(scores):.1f}")
ax.set_title("Distribution of Student Scores")
ax.set_xlabel("Score"); ax.set_ylabel("Count"); ax.legend()
plt.tight_layout(); plt.show()
````

## 5. Scatter Plot


````python
exp = np.random.randint(0, 30, 150)
sal = 30000 + exp*2000 + np.random.normal(0, 8000, 150)

fig, ax = plt.subplots(figsize=(7,4))
ax.scatter(exp, sal, alpha=0.5, color="steelblue")
z = np.polyfit(exp, sal, 1)
ax.plot(np.linspace(0,30), np.poly1d(z)(np.linspace(0,30)), "r--", label="Trend")
ax.set_title("Experience vs Salary")
ax.set_xlabel("Experience (years)"); ax.set_ylabel("Salary ($)")
ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
````

## 6. Subplots — Multiple Charts in One Figure


````python
fig, axes = plt.subplots(2, 2, figsize=(11, 7))
fig.suptitle("Employee Dashboard", fontsize=15, fontweight="bold")

ages = np.random.randint(22, 60, 200)
salaries = np.random.normal(60000, 15000, 200)
perf = np.random.choice([1,2,3,4,5], 200)
gender = np.random.choice(["Male","Female"], 200)

axes[0,0].hist(ages, bins=15, color="skyblue", edgecolor="white")
axes[0,0].set_title("Age Distribution")

axes[0,1].hist(salaries, bins=20, color="salmon", edgecolor="white")
axes[0,1].set_title("Salary Distribution")

counts = pd.Series(perf).value_counts().sort_index()
axes[1,0].bar(counts.index, counts.values, color="mediumseagreen")
axes[1,0].set_title("Performance Ratings")

gc = pd.Series(gender).value_counts()
axes[1,1].pie(gc, labels=gc.index, autopct="%1.0f%%", colors=["steelblue","lightcoral"])
axes[1,1].set_title("Gender")

plt.tight_layout(); plt.show()
````

## ✅ Summary

| Chart | Use | Function |

| Bar | Compare categories | `ax.bar()` |
| Histogram | Distribution | `ax.hist()` |
| Scatter | Relationship | `ax.scatter()` |
| Multiple | Dashboards | `plt.subplots(rows,cols)` |

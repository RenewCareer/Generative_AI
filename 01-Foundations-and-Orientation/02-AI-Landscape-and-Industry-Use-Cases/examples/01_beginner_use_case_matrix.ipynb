# Beginner: Load and Query a GenAI Use-Case Dataset

**Sub-topic 1.2 — AI Landscape and Industry Use Cases**

This notebook teaches the load-filter-aggregate pattern you'll reuse constantly: read structured data into a `pandas` DataFrame, filter rows by condition, and summarize with group-by counts.

```python
from pathlib import Path

import pandas as pd

DATA_PATH = Path("industry_use_cases.csv")
df = pd.read_csv(DATA_PATH)

print(f"Loaded {len(df)} GenAI use cases across {df['industry'].nunique()} industries")
df.head(5)
```

## Query 1 — all use cases for the Healthcare industry

```python
healthcare = df[df["industry"] == "Healthcare"]
healthcare[["use_case", "genai_capability", "modality"]]
```

## Query 2 — all text-modality use cases that are still "Emerging"

```python
emerging_text = df[(df["modality"] == "text") & (df["maturity"] == "Emerging")]
emerging_text[["industry", "use_case"]]
```

## Query 3 — how many use cases per modality?

```python
df["modality"].value_counts()
```

## Query 4 — which industry has the most documented use cases?

```python
counts = df.groupby("industry").size().sort_values(ascending=False)
print(counts)
print(f"\nTop industry: {counts.idxmax()} ({counts.max()} use cases)")
```
